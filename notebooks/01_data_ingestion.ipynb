```python
def load_dataset(filename):
    filepath = os.path.join(RAW_DATA_PATH, filename)

    if not os.path.exists(filepath):
        raise FileNotFoundError(f"File not found: {filepath}")

    df = pd.read_csv(filepath)
    print(f" Loaded {filename}: {df.shape[0]} rows, {df.shape[1]} columns")

    return df 
```

```python
df_hourly = load_dataset("saleshourly.csv")
```

```python
# Load all datasets using our function
df_hourly  = load_dataset("saleshourly.csv")
df_daily   = load_dataset("salesdaily.csv")
df_weekly  = load_dataset("salesweekly.csv")
df_monthly = load_dataset("salesmonthly.csv")
```

```python
def profile_dataset(df, name):
    """
    Print basic profile of dataframe.

    Args:
        df (pd.DataFrame): Dataset to profile
        name (str): Name label for display
    """
    print(f"\n{'='*50}")
    print(f"DATASET: {name}")
    print(f"{'='*50}")
    print(f"Shape :{df.shape}")
    print(f"Date range : {df['datum'].min()} -> {df['datum'].max()}")
    print(f"Null Values: {df.isnull().sum().sum()}")
    print(f"Duplicates: {df.duplicated().sum()}")
    print(f"{'='*50}")

#Profile all datasets
for name,df in datasets.items():
    profile_dataset(df, name)
```

```python
for name,df in datasets.items():
    zeros = (df[['M01AB', 'M01AE',	'N02BA',	'N02BE',	'N05B',	'N05C',	'R03',	'R06']] == 0).sum().sum()
    print(f"{name}: {zeros} zero values in drug columns")
```

```python
drug_cols = ['M01AB', 'M01AE',	'N02BA',	'N02BE',	'N05B',	'N05C',	'R03',	'R06']

all_zero_rows = df_hourly[(df_hourly[drug_cols]==0).all(axis=1)]
print(f"Rows where ALL drugs are zero: {len(all_zero_rows)}")
print(all_zero_rows[['datum','Hour','Weekday Name']].head(20))
```

```python
zero_by_hour = all_zero_rows.groupby('Hour').size().sort_index()
print(zero_by_hour)
```

```python
zero_by_weekday = all_zero_rows.groupby('Weekday Name').size().sort_index(ascending=False)
print(zero_by_weekday)
```
